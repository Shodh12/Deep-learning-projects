```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
```

```python
#loading rthe data from sklearn
dataset=datasets.load_breast_cancer()
```

```python
#loading the data innto the dataframe
data_frame=pd.DataFrame(dataset.data,columns=dataset.feature_names)
```

```python
#adding the target column to the dataframe
data_frame['label']=dataset.target
```

```python
data_frame.groupby('label').sum()
```

```python
#splitting the data into features and target
X=data_frame.drop(columns='label',axis=1)
Y=data_frame['label']
```

```python
#spliting the data into training and testing
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train_std=scaler.fit_transform(X_train)

X_test_std=scaler.transform(X_test)
```

Building up the neural network

```python
#importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)  #to get constant result everytime we run
from tensorflow import keras
```

```python
#setting the layes in neural network

model = keras.Sequential ([
                          keras.layers.Flatten(input_shape=(30,)),
                          keras.layers.Dense(20, activation ='relu'),
                          keras.layers.Dense(2, activation = 'sigmoid')
])
```

```python
#compiling the neural network

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',     #since it has no one-not-encoding[1,0,0]or[0,1,0]or[0,0,1] here we have only 2 labels 0 or 1
              metrics=['accuracy'])
```

```python
#training the neural network
history= model.fit(X_train_std, Y_train, validation_split=0.1, epochs=10)
```

Visualizing the accuracy and loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('model accuarcy')
plt.ylabel('accuarcy')
plt.xlabel('epoch')

plt.legend(['training data', 'validation_data'], loc="lower right")
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data', 'validation_data'], loc="upper right")
```

```python
#Accuracy on the test data
loss, accuracy= model.evaluate(X_test_std, Y_test)
```

```python
input_data=(17.99,10.38,122.8,1001,0.1184,0.2776,0.3001,0.1471,0.2419,0.07871,1.095,0.9053,8.589,153.4,0.006399,0.04904,0.05373,0.01587,0.03003,0.006193,25.38,17.33,184.6,2019,0.1622,0.6656,0.7119,0.2654,0.4601,0.1189)

#converting to array
input_data_as_array = np.asarray(input_data)

#reshaping
input_data_reshape=input_data_as_array.reshape(1, -1)

#standardizing the input_data
input_data_std=scaler.transform(input_data_reshape)

prediction = model.predict(input_data_std)
print(prediction)

prediction_label = [np.argmax(prediction)]
print(prediction_label[0])

if(prediction_label[0]==0):
  print('The breast cancer is Malignant')
else:
  print('The breast cancer is Benign')
```
